# Ground truth dataset acquisition notebook

**Description**

This notebook downloads ground truth datasets from Dresden's Open Data Portal to explore and create ground truth data for Public Urban Green Spaces (PUGS) detection. It retrieves multiple datasets including parks, green spaces, and leisure areas. These datasets will be used for training and evaluating PUGS detection models in later steps.

For **Urban Atlas Land Cover/Land use 2018 or European Urban Atlas (EUA) dataset**, User need to **download it manually** from the link provided.

**Overview of datasets**

| Data source     | Geometry type     | Dataset name     | Description      |
|-----------------|-------------------|------------------|------------------|
|  Copernicus Land Monitoring Service | Polygon/Multi-polygon | [Urban Atlas Land Cover/Land use 2018](https://land.copernicus.eu/en/products/urban-atlas/urban-atlas-2018) <br> (European Urban Atlas or EUA) <br><br> (Temporal extent of dataset: 2017-2019) | Provide land cover and land use data in Functional Urban Areas (FUA) in Europe and UK. The dataset consists of 17 urban classes with the Minimum Mapping Unit (MMU) of 0.25 ha and for 10 rural classes with the MMU of 1 ha. <br><br>**Note:** this dataset is updated every 6 years. |
| Dresden Open Data Portal  <br> (Origin from Office for Urban Green Spaces and Waste Management) | Polygon/Multi-polygon | [Park- und Grünanlagen (Flächen)](https://kommisdd.dresden.de/net3/public/ogc.ashx?Service=IKX&NodeId=751&RenderHint=TargetHTML) <br> (Park and green spaces)  <br><br> (Last update: 2020) | Provide the areas of parks and green spaces within Dresden. It includes **only green spaces that are under the administration of the state capital Dresden**. The dataset includes the areas that fall into: <br> • I1 - parks <br> • F41 - Green city square (Begrünte Stadtplätze) <br> • F43 - Other green and open spaces (Sonstige Grün- und Freiflächen) <br><br> [More details of each category](https://kommisdd.dresden.de/net3/public/ogcsl.ashx?Service=IKX&Request=GetBlobData&ts=638522102964126790&StObjectDataId=31389&ikxStoreId=) |
| Dresden Open Data Portal  <br> (Origin from Office for Geodata and Cadastre) | Polygon/Multi-polygon | [Grün- und Freiflächen](https://kommisdd.dresden.de/net3/public/ogc.ashx?Service=IKX&NodeId=1638&RenderHint=TargetHTML) <br> (Green and open spaces) <br><br> (Last update: 2024) | Provide the areas of green and open spaces in Dresden. The dataset only includes the areas that fall into: <br> • F - Green and open spaces (Grün- und Freifläche) <br> • K - Cemeteries (Friedhöfe) <br> • L - Forest/Woodland (Wald / Gehölz) <br><br> [More details of each category](https://kommisdd.dresden.de/net3/public/ogcsl.ashx?Service=IKX&Request=GetBlobData&ts=638522102964126790&StObjectDataId=31389&ikxStoreId=) |
| Dresden Open Data Portal <br> (Origin from Environment Agency) | Polygon/Multi-polygon | Erholungs- und Freizeitflächen (Recreational and leisure areas) <br><br> (Last update: 2021) | Provide the areas that fall into recreational areas according to different types of land uses. The whole dataset is [Nutzungsarten (Strukturtypen) – Kartierung](https://kommisdd.dresden.de/net3/public/ogc.ashx?Service=IKX&NodeId=1059&RenderHint=TargetHTML) <br> (Types of use - structure types - Mapping), but with the capability of WFS, specific layer can be selected to download. <br><br> [More details of each category](https://kommisdd.dresden.de/net3/public/ogcsl.ashx?Service=IKX&Request=GetBlobData&ts=638522102964126790&StObjectDataId=31389&ikxStoreId=) |
| Dresden Open Data Portal <br> (Origin from Environment Agency) | Point | [Parkanlagen, Gärten](https://kommisdd.dresden.de/net3/public/ogc.ashx?Service=IKX&NodeId=138&RenderHint=TargetHTML) (Parks, gardens) <br><br> (Last update: 2021) | Provide the location of all parks and park-like gardens within the urban area of Dresden. This dataset contains not only parks that are under the administration of the state capital Dresden but also parks under other organizations. <br><br> [More details](https://www.dresden.de/de/stadtraum/umwelt/gruenes-dresden/gruenanlagen-parks.php) |

```python
import os
import sys

# Add the root directory to Python path so that we can import the modules
root_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.append(root_dir)
```

```python
import geopandas as gpd
from pugs_detection.utils import print_basic_info
from pugs_detection.ground_truth import download_ground_truth_data
```

# 2. Download Urban Atlas Land Cover/Land use 2018

User need to manually download this dataset from https://land.copernicus.eu/en/products/urban-atlas/urban-atlas-2018#download .
Next, user can select **Download pre-packaged data collections** and search **Dresden** to download the data.

**Requirement:** User need to have an EU Login account.

# 3. Download Green and Open Spaces

```python
url = "https://kommisdd.dresden.de/net4/public/ogcapi/collections/L1266/items"
green_and_os_gdf = download_ground_truth_data(url=url, og_crs=4326, new_crs=32633)
```

```python
geojson_output_path = "../data/raw/ground truth/green_and_openspaces_dataset.geojson"
green_and_os_gdf.to_file(geojson_output_path, driver="GeoJSON")
```

```python
green_and_os_gdf = gpd.read_file(geojson_output_path)
```

```python
print_basic_info(green_and_os_gdf)
```

# 4. Download Parks and greenspaces dataset

```python
url = "https://kommisdd.dresden.de/net4/public/ogcapi/collections/L569/items"
park_and_gs_gdf = download_ground_truth_data(url=url, og_crs=4326, new_crs=32633)
```

```python
geojson_output_path = "../data/raw/ground truth/parks_and_greenspaces_dataset.geojson"
park_and_gs_gdf.to_file(geojson_output_path, driver="GeoJSON")
```

```python
park_and_gs_gdf = gpd.read_file(geojson_output_path)
```

```python
print_basic_info(park_and_gs_gdf)
```

# 5. Download Parks (I type) dataset

```python
# WFS service URL
url = "https://kommisdd.dresden.de/net3/public/ogcsl.ashx"

# WFS parameters - removed outputFormat parameter since XML is the only option
params = {
    "NODEID": "1059",
    "Service": "WFS",
    "typenames": "cls:L1263",
    "request": "GetFeature",
    "version": "2.0.0",  # Using standard WFS 2.0.0
}

leisure_gdf = download_ground_truth_data(
    url=url, params=params, og_crs=25833, new_crs=32633
)
```

```python
geojson_output_path = "../data/raw/ground truth/leisure_area_dataset.geojson"
leisure_gdf.to_file(geojson_output_path, driver="GeoJSON")
```

```python
leisure_gdf = gpd.read_file(geojson_output_path)
```

```python
print_basic_info(leisure_gdf)
```
